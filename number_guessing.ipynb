```python
# A simple number guessing game using python.
import random # For random numbers

def number_guessing():
    number = random.randint(1,100)
    while True:

        try:
            guess = int(input("Guess a number from 1 to 100: "))

            if number == guess:
                print("Congratulations! You guessed the number!")
                break
            elif number > guess:
                print("Too low!")
            else:
                print("Too high!")
            

        except ValueError:
            print("Enter a valid number")

number_guessing()
```


## 逐行详细解释：

**第1行：`import random`**
- **import** 是"导入"的意思，就像从工具箱里拿工具

- 这行代码让我们可以使用生成随机数的功能
- **初学者易错**：忘记导入模块，或者拼写错误

**第3行：`def guess_number_game():`**
- **def** 是"定义函数"的关键字，就像给一套操作起个名字
- **guess_number_game** 是函数名（你可以起任何名字）
- **()** 里面可以放参数，这里是空的
- **冒号(:)** 表示"接下来的内容属于这个函数"
- **语法规则**：冒号后面必须换行并缩进
- **初学者易错**：忘记写冒号，或者函数名用中文

**第4行：`secret_number = random.randint(1, 100)`**
- 这行有**4个空格的缩进**（属于函数内部）
- **=** 是赋值符号，把右边的值给左边的变量
- **random.randint(1, 100)** 生成1到100之间的随机整数
- **初学者易错**：缩进不对，或者不理解赋值和相等的区别

**第5行：`attempts = 0`**
- 创建一个变量记录猜测次数，初始值是0
- **初学者易错**：变量名用中文或包含空格

**第7-8行：`print()` 语句**
- **print()** 用来在屏幕上显示文字
- 引号里的内容会原样显示
- **初学者易错**：忘记引号，或者中英文引号混用

**第10行：`while True:`**
- **while** 是"当...的时候"的意思，创建循环
- **True** 表示条件永远成立，所以是无限循环
- **冒号** 表示循环体开始
- **初学者易错**：忘记冒号，或者不理解无限循环的概念

**第11行：`user_input = input("请输入你的猜测：")`**
- **input()** 让用户输入内容
- 括号里的文字是提示信息
- 用户输入的内容存储在user_input变量里
- **初学者易错**：不知道input()返回的是字符串类型

**第12行：`attempts += 1`**
- **+=** 是简写，等同于 `attempts = attempts + 1`
- 每次猜测后，次数加1
- **初学者易错**：写成 `attempts + 1` 而忘记赋值

**第14行：`if user_input.isdigit():`**
- **if** 是"如果"的意思，用来做判断
- **isdigit()** 检查字符串是否全是数字
- **冒号** 表示条件成立时要执行的代码块开始
- **初学者易错**：忘记冒号，或者不理解字符串方法

**第15行：`guess = int(user_input)`**
- **int()** 把字符串转换成整数
- 因为input()返回的是字符串，需要转换才能比较大小
- **初学者易错**：直接用字符串比较数字大小

**第17行：`if guess == secret_number:`**
- **==** 是"相等"的意思（两个等号！）
- **单个=** 是赋值，**双个==** 是比较
- **初学者易错**：用单个=做比较，这是最常见的错误！

**第18-21行：猜对的情况**
- **f"..."** 是格式化字符串，{}里可以放变量
- **break** 跳出循环
- **初学者易错**：不理解f字符串的语法

**第22行：`elif guess < secret_number:`**
- **elif** 是"否则如果"的意思
- **<** 是"小于"的比较符号
- **初学者易错**：写成else if，或者搞混比较符号

**第25行：`else:`**
- **else** 是"否则"的意思，处理其他所有情况
- **初学者易错**：逻辑不清楚，不知道什么时候用else

## 关键语法规则总结：

1. **缩进规则**：Python用缩进表示代码层级，通常用4个空格
2. **冒号规则**：if、while、def、else等后面必须加冒号
3. **比较符号**：= 是赋值，== 是相等比较
4. **字符串**：用引号包围，可以用单引号或双引号

## 初学者最容易犯的错误：

1. **缩进错误**：不对齐或混用空格和Tab键
2. **忘记冒号**：if、while、def后面忘记加:
3. **= 和 == 搞混**：赋值和比较不分
4. **引号错误**：中英文引号混用，或者忘记配对
5. **变量名错误**：用中文、空格或Python关键字做变量名
